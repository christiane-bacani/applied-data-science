### **Lesson: Basic Plots with Matplotlib**

Objective: Use functions from `pyplot` sub-package of `matplotlib` package to perform data visualization such as **Line Chart** or **Scatter Plot**.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../Datasets/Fifa.csv")
```

```python
age = df["Age"]
total_stats_score = df["Total_Stats Score"]

plt.plot(age, total_stats_score)
plt.xlabel("Age")
plt.ylabel("Total Statistical Score")
plt.show()
```

```python
plt.scatter(age, total_stats_score)
plt.xlabel("Age")
plt.ylabel("Total Statistical Score")
plt.show()
```
