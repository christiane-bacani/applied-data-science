### **Lesson: Basic Plots with Matplotlib**

**Objective:** Use functions from `pyplot` sub-package of `matplotlib` package to perform data visualization methods such as [**Line Chart**](https://www.cuemath.com/data/line-chart/) to represent the changes of data-points that are connected through line over a series of time and  [**Scatter Plot**](https://www.geeksforgeeks.org/maths/scatter-plot/) to visualize the correlation or relationship of two variables.


What I learned:
- Import `pyplot` sub-package from the main package `matplotlib` to use different methods or functions to perform data visualization.
- Use `plot()` function to perform **Line Chart** visualization to track the changes of different data points over a series of time.
- Use `scatter()` function to perfoorm **Scatter Plot** visualization to display the correlation or relationship between two variables.
- Use `xlabel()` and `ylabel()` to create header for the horizontal-axis and vertical-axis of the line chart and scatter plot.
- Use `show()` function to display the plot we created.

#### **Setup environment**

Setup the packages, sub-packages, and modules to use to perform Data Visualization.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("..\Datasets\global_natural_disasters_2000_2025.csv")
```

### **Line Chart**

Use line chart to visualize the changes of earthquake magnitude over a course of time using `plot()` function.

```python
# Clean the variables before plotting
df["date"] = pd.to_datetime(df["date"])
df = df[df["date"].dt.year > 2019]

# Visualize the trend/changes using line-plot
plt.plot(
    df["date"],
    df["magnitude"]
)
plt.xlabel("Date")
plt.ylabel("Magnitude")
plt.show()
```

As you can see above, the chart visualize the changes of magnitude over the course of time. We can see the highest earthquake magnitude happen on the year `2025`.

### **Scatter Plot**

Use scatter plot to visualize the correlations or relationships between two variables using `scatter()` function.

```python
# Visualize the correlation/relationship using scatter-plot
plt.scatter(
    df["date"],
    df["magnitude"]
)
plt.xlabel("Date")
plt.ylabel("Earthquake Magnitude")
plt.show()
```

As you can see, the chart visualize the relationship between the variable `Date` and the variable `Earthquake Magnitude`. Based on the chart, we conclude that there's no correlation between the two variables.
