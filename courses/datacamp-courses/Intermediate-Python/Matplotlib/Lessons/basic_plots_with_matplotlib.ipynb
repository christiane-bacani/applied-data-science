### **Chapter: Matplotlib**
### **Lesson #1: Basic plot with Matplotlib**

---

**Data visualization** is really important in *Data Analysis* because it enables us to explore our dataset and gain valuable insights. Once we found the insights, again, we need visualization to be able to share our valuable insights with other people.

There are many visualization packages in **Python**, but the mother of them all, is [matplotlib](https://matplotlib.org/).
