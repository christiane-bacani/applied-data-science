### **Chapter: Matplotlib**
### **Lesson #1: Basic plot with Matplotlib**

---

**Data visualization** is really important in *Data Analysis* because it enables us to explore our dataset and gain valuable insights. Once we found the insights, again, we need visualization to be able to share our valuable insights with other people.

There are many visualization packages in **Python**, but the mother of them all, is [matplotlib](https://matplotlib.org/).

To use matplotlib, we need to select the subpackage `pyplot` and use an alias `plt` for a better convention of name.

```python
import matplotlib.pyplot as plt
```

- `plot()`: A function from sub-package `pyplot` of the package `matplotlib` that tells **Python** on *what* to plot and *how* to plot it by using [line chart](https://www.investopedia.com/terms/l/linechart.asp).
- `show()`: A function to actually displays the plot.

```python
year = [1950, 1970, 1990, 2010]
population = [2.519, 3.692, 5.263, 6.972]

# plt.plot(horizontal-axis, vertical-axis)
plt.plot(year, population)

# We use `show()` function to display the plot we created, because sometimes we add more
# ingredients to our plot before displaying it.
plt.show()
```

- `scatter()`: A function from sub-package `pyplot` of the package `matplotlib` that tells **Python** on *what* to plot and *how* to plot it by using a [scatter plot](https://www.geeksforgeeks.org/maths/scatter-plot/).

```python
# plt.scatter(horizontal-axis, vertical-axis)
plt.scatter(year, population)
plt.show()
```
