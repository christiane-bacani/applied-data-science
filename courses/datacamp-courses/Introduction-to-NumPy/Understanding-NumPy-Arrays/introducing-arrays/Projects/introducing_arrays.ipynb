```python
import pandas as pd 
import numpy as np

df = pd.read_csv("../Datasets/mental_health.csv")
```

Creating one-dimensional array.

```python
person_id = np.array(df["Person_ID"])
age = np.array(df["Age"])
```

Creating two-dimensional array.

```python
# Row-wise
row_wise_arr = np.array([person_id, age])
print(row_wise_arr)
print()

# Column-wise
column_wise_arr = np.column_stack([person_id, age])
print(column_wise_arr)
```

```python
three_d_arr = np.array([
    [person_id[:1000],age[:1000]], # First matrix
    [person_id[1000:],age[1000:]] # Second matrix
])
print(three_d_arr)
```
