### **Lesson: Introducing Arrays**

Building **NumPy Arrays** in different methods such as building **NumPy Array** using **Pandas** library or building from scratch using different methods such as `np.zeros()`, `np.random.random()`, and `np.arange()`.

#### **Setup environment**

Setup the libraries and modules to use to create *NumPy Arrays* by running the code snippet below.

```python
import pandas as pd
import numpy as np
from numpy import ndarray

df = pd.read_csv("../Datasets/mental_health.csv")
```

#### **Create 1D Array**

Create 1D Array using Pandas Library by extracting or selecting an example of a Series from the given DataFrame.

```python
one_dim_array = np.array(df["Person_ID"])
print("1D Array:")
print(one_dim_array)
```

### **Create 2D Array**

Create 2D Array using `np.column_stack()` method.

```python
two_dim_array = np.column_stack([df["Person_ID"], df["Age"]])
print("2D Array:")
print(two_dim_array)
```

### **Create 3D Array**

Create 3D Array by combining the methods: `np.array()` and `np.column_stack()`

```python
person_id1 = df["Person_ID"][:1000]
occupation1 = df["Occupation"][:1000]

person_id2 = df["Person_ID"][1000:]
occupation2 = df["Occupation"][1000:]

three_dim_array = np.array([
    np.column_stack([person_id1, occupation1]),
    np.column_stack([person_id2, occupation2])
])

print("3D Array:")
print(three_dim_array)
```

### **Create 1D Array using `np.zeroes()`**

Create 1D Array with `0` elements by using `np.zeroes()`.

```python
def create_zero_elem_arr(data_shape: tuple) -> ndarray:
    """Create NumPy Array based on the given dataset shape."""
    zero_elem_arr = np.zeros(shape=data_shape)
    return zero_elem_arr

one_dim_zero_elem_array = create_zero_elem_arr(data_shape=(100))
print("1D Array using np.zeros(100):")
print(one_dim_zero_elem_array)
```

### **Create 2D Array using `np.zeroes()`**

Create 2D Array with `0` elements by using `np.zeroes()`.

```python
two_dim_zero_elem_array = create_zero_elem_arr(data_shape=(100, 4))
print("2D Array using np.zeroes(100, 4):")
print(two_dim_zero_elem_array)
```

### **Create 3D Array using `np.zeroes()`**

Create 3D Array with `0` elements by using `np.zeroes()`.

```python
three_dim_zero_elem_array = create_zero_elem_arr(data_shape=(3, 10, 5))
print("3D Array using np.zeroes(3, 10, 5):")
print(three_dim_zero_elem_array)
```

### **Create 1D Array using np.random.random()**

Create 1D Array with random elements using `np.random.random()`.

```python
def create_random_elem_arr(data_shape: tuple) -> ndarray:
    """Create NumPy Array with random elements based on the given dataset shape"""
    random_elem_arr = np.random.random(size=data_shape)
    return random_elem_arr

one_dim_random_elem_array = create_random_elem_arr(data_shape=(50))
print("1D Array using np.random.random(50):")
print(one_dim_random_elem_array)
```

### **Create 2D Array using np.random.random()**

Create 2D Array with random elements using `np.random.random()`.

```python
two_dim_random_elem_array = create_random_elem_arr(data_shape=(12, 4))
print("2D Array using np.random.random(12, 4):")
print(two_dim_random_elem_array)
```

### **Create 3D Array using np.random.random()**

Create 3D Array with random elements using `np.random.random()`.

```python
three_dim_random_elem_array = create_random_elem_arr(data_shape=(2, 50, 3))
print("3D Array using np.random.random(2, 50, 3):")
print(three_dim_random_elem_array)
```

```python
def create_range_elem_arr(data_shape: tuple) -> ndarray:
    if type(data_shape) is int:
        range_elem_arr = np.arange(data_shape)

    elif len(data_shape) == 2:
        range_elem_arr = np.arange(data_shape[0], data_shape[1])

    elif len(data_shape) == 3:
        range_elem_arr = np.arange(data_shape[0], data_shape[1], data_shape[2])

    else:
        raise Exception("Invalid parameters!")

    return range_elem_arr

one_dim_range_elem_array = create_range_elem_arr(data_shape=(51))
print("1D Array using np.arange(51):")
print(one_dim_range_elem_array)
```
