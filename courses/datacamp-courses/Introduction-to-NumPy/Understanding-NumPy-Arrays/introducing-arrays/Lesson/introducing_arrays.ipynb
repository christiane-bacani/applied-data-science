**NumPy** is the core library for scientific computing in Python. Different libraries such as *Sci-Py*, *Matplotlib*, and *Pandas* are built in top of **NumPy API**. Even popular machine learning libraries such as *Scikit-learn* and *Tensorflow* uses **NumPy arrays** as inputs. Any person or developers who uses numbers for different types of computations will come across NumPy Arrays.

**Array** is the main object of NumPy and stores data in a grid-like structure that is similar in spreadsheets or tables. Arrays can have any number of dimensions and each dimensions can have any length.

Creating one-dimensional array in NumPy.

```python
import numpy as np

city = ["Charlotte", "New York", "Orlando", "Boston", "Washington", "Cleveland"]
one_d_ar = np.array(city)
print(one_d_ar)
```

Creating two-dimensional array.

```python
city_and_team = [
    ["Boston", "Celtics"],
    ["Brooklyn", "Nets"],
    ["New York", "Knicks"],
    ["Philadelphia", "76ers"],
    ["Toronto", "Raptors"]
]
two_d_arr = np.array(city_and_team)
print(two_d_arr)
```

Creating three-dimensional array.

```python
city_team_conf = [
    [
        ["Boston", "Celtics", "East"],
        ["Brooklyn", "Nets", "East"],
        ["New York", "Knicks", "East"],
        ["Philadelphia", "76ers", "East"],
        ["Toronto", "Raptors", "East"]
    ],
    [
        ["Utah", "Jazz", "West"],
        ["Golden State", "Warriors", "West"],
        ["Los Angeles", "Clippers", "West"],
        ["Los Angeles", "Lakers", "West"],
        ["Phoenix", "Suns", "West"]
    ]
]
three_d_arr = np.array(city_team_conf)
print(three_d_arr)
```

The difference of Python list and NumPy array is how they stored values. Python list stores values or datas with different datatype (**heterogenous**) in the same dimension while NumPy array stores values or datas with the same datatype (**homogenous**).

```python
# Python list
nba_player = ["Nikola Jokić", "Denver Nuggets", "Center", 211, 128]
print(nba_player)

# NumPy Array
nba_player = np.array(["Nikola Jokić", "Denver Nuggets", "Center", 211, 128])
print(nba_player)
```

In NumPy, we can create arrays by passing a data stored in list as an argument for `np.array()` or create from scratch using different NumPy functions such as `np.zeros()`, `np.random.random()`, and `np.arrange()`.

Creating arrays from scratch using `np.zeros()`

```python
# Create array with 1 row and 5 columns of zero elements
print(np.zeros(5))
print()

# Crate array with 9 rows and 4 columns of zero elements
zero_arr = np.zeros((9, 4))
print(zero_arr)
```

Creating arrays from scratch using `np.randon.random()`

> **Note**: `np.random.random()` generates random float numbers ranging from 0 to 1 (not any numbers). Also, `random()` is a function from **random** module that's why when calling the function we specify it as `np.random.random()`.

```python
# Generate random single value if shape are not specified
print(np.random.random())
print()

# Generate array with 1 row and 5 columns with random numbers
print(np.random.random((5)))
print()

# Generate array with 9 rows and 4 columns with random numbers
random_arr = np.random.random((9, 4))
print(random_arr)
print()
```

Creating arrays from scratch using `np.arange()`

```python
# Creates a sequence of numbers in one dimension starting from 0 to 5 (excluded)
print(np.arange(5))

# Creates a sequence of numbers in one dimension starting from -5 to 5 (exclusive)
print(np.arange(-5, 5))

# Creates a sequence of numbers in one dimension starting from 10 to 0 (exclusive)
# and increment by -2
print(np.arange(10, 0, -2))

# Results in empty array because first argument >= to the second argument
print(np.arange(5, 5))
```

We can combine different functions to create array from scratch.

```python
random_arr = np.random.random((4))
first_range_arr = np.arange(4)
second_range_arr = np.arange(4, 0, -1)

combined_arr = np.column_stack([first_range_arr, second_range_arr])
```

`np.arange()` is really useful for plotting!

```python
from matplotlib import pyplot as plt

plt.scatter(
    np.arange(1, 11),
    np.arange(1, 11)
)
plt.show()
```
